# Identify and handle missing values

## Check missing values

```python
# Check Missing values
df.isnull().sum()
```

#### Observation
- Missing values were found in the Job (148), Industry (3), Highly_rated_for (1), and Location (2) columns.
- The remaining columns contain no missing values and are complete.

## Handling missing values

```python
# fill missing values in Industry column
df["Job"] = df["Job"].fillna(df["Job"].median())
```

```python
# fill missing values in Industry column
df["Industry"] = df["Industry"].fillna(df["Industry"].mode()[0])
```

```python
# fill missing values in Highly_rated_column column
df["Highly_rated_for"] = df["Highly_rated_for"].fillna("Unknown")
```

```python
# fill missing values in Location column
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
```

```python
# check missing values again
df.isnull().sum()
```

#### Obseravation
- Missing values were successfully handled using appropriate techniques: median for the Job column, mode for Industry and Location, and "Unknown" for the Highly_rated_for column.
- After treatment, all columns contain 0 missing values, making the dataset complete and ready for further analysis.

# Bivariate Analysis

## Average Salary by Top 10 Industries (GroupBy + Bar Plot)
### Continuous + Categorical Variables (Pivot Table)

```python
salary_by_industry = (df.groupby("Industry")["Salary"].mean().sort_values(ascending=False).head(10))
plt.figure(figsize=(10,5))
sns.barplot(x=salary_by_industry.index,y=salary_by_industry.values,palette="Blues_d")

plt.title("Top 10 Industries by Average Salary")
plt.xlabel("Industry")
plt.ylabel("Average Salary")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#### Observation

- Average salary differs across industries, showing variations in employee compensation.
- Some industries offer noticeably higher average salaries than others.

## Average Salary by Industry and Location (Pivot Table)
### Continuous + Categorical Variables (Pivot Table)

```python
pivot = pd.pivot_table(df,values="Salary",index="Industry",columns="Location",aggfunc="mean")
plt.figure(figsize=(9,6))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("Average Salary by Industry and Location")
plt.show()
```

#### Observation

- Average salary varies across different industry and location combinations.
- The pivot table helps identify high-paying industries in specific locations.

## Industry vs Highly Rated For (Crosstab)
### Categorical → Categorical (Crosstab)

```python
cross = pd.crosstab(df["Industry"], df["Highly_rated_for"])

plt.figure(figsize=(9,6))
sns.heatmap(cross, cmap="Blues")
plt.title("Industry vs Highly Rated For")
plt.xlabel("Highly Rated For")
plt.ylabel("Industry")
plt.show()
```

#### Observation
- Employee preferences for workplace benefits vary across different industries.
- Some industries are more frequently associated with specific highly rated factors.

# Multivariate Analysis

```python
# Correlation Heatmap

corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap="cubehelix",linewidths=0.5,fmt=".2f")
plt.title("Correlation Heatmap",fontsize=16,fontweight="bold")
plt.show()
```

#### Observation
- The correlation heatmap shows the strength and direction of relationships among the numerical variables in the dataset.
- Review and Salary exhibit a relatively stronger positive correlation, while Rating has a weak correlation with the other numerical variables.

```python
# Review vs Salary vs Rating

plt.figure(figsize=(10,6))
scatter = plt.scatter(df["Review"],df["Salary"],c=df["Rating"],cmap="plasma",alpha=0.7)
plt.colorbar(scatter, label="Rating")
plt.xlabel("Review Count", fontsize=12)
plt.ylabel("Salary", fontsize=12)
plt.title("Review vs Salary vs Rating",fontsize=16,fontweight="bold")
plt.grid(linestyle="--", alpha=0.4)
plt.show()
```

#### Observation
- The scatter plot indicates that companies with a higher number of reviews generally tend to offer higher salaries, although the relationship is not perfectly linear.
- Company ratings remain relatively consistent across different review counts and salary levels, suggesting that higher salaries do not always correspond to higher ratings.

```python
# Average Salary by Industry and Location

pivot = df.pivot_table(values="Salary",index="Industry",columns="Location",aggfunc="mean")
plt.figure(figsize=(10,6))
sns.heatmap(pivot,cmap="inferno",annot=False,linewidths=0.5)
plt.title("Average Salary : Industry vs Location",fontsize=16,fontweight="bold")
plt.xticks(rotation=45)
plt.show()
```

#### Observation
- The heatmap highlights that average salaries vary across different industry and location combinations, indicating regional differences in compensation.
- Some industry–location pairs show higher average salaries than others, suggesting that both industry and location influence salary levels.

```python
# Pairplot for Numerical Features

sns.pairplot(df[["Rating","Salary","Review","Job"]],diag_kind="kde")
plt.suptitle("Pairplot of Numerical Features",y=1.02,fontsize=16,fontweight="bold")
plt.show()
```

#### Observation
